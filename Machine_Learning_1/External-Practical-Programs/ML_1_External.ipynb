**1. Simple Linear Regression – House price based on area**

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Area in square feet and price in lakh

# Input Feature (X) is in 2D format
X = np.array([[500], [700], [900], [1100], [1300]])

# Target Value (y) is in 1D format
y = np.array([5, 7, 9, 11, 13])

# Create a Linear Regression model
model = LinearRegression()

# Train the model using input and target values
model.fit(X, y)

# Area for which we want to predict the price
area = [[1000]]

# [0] accesses the first predicted value from the array
print("Predicted house price:", model.predict(area)[0], "lakh")

# [0] accesses the first slope value (only one input feature)
print("Slope:", model.coef_[0])

# Displays the intercept value (no index needed)
print("Intercept:", model.intercept_)
```

**3. Multivariate Linear Regression – House price**

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create a dataset using Pandas DataFrame
data = pd.DataFrame({
    "Area": [600, 800, 1000, 1200, 1500],
    "Bedrooms": [1, 2, 2, 3, 3],
    "Age": [10, 8, 5, 4, 2],
    "Price": [30, 42, 55, 68, 85]
})

# Select input features (X)
X = data[["Area", "Bedrooms", "Age"]]

# Select target value (y)
y = data[["Price"]]

# Create a Multiple Linear Regression model
model = LinearRegression()

# Train the model using the dataset
model.fit(X, y)

# Predict price for a new house
new_house = pd.DataFrame({
    "Area": [1200],
    "Bedrooms": [4],
    "Age": [6]
})

# [0] accesses the first predicted value
print("Predicted price:", model.predict(new_house)[0])
```

**4. Multivariate Linear Regression – Employee salary**

```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Create a dataset using Pandas DataFrame
data = pd.DataFrame({
    "Experience": [2, 3, 4, 5, 6],
    "Education": [9, 12, 16, 18, 20],
    "Age": [25, 30, 35, 40, 45],
    "Salary": [50000, 60000, 70000, 80000, 90000]
})

# Select input features (X)
X = data[["Experience", "Education", "Age"]]

# Select target value (y)
y = data["Salary"]

# Create a Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Create data for a new employee
new_employee = pd.DataFrame({
    "Experience": [5],
    "Education": [15],
    "Age": [45]
})

# Predicts Salary for new Employee
print(f"Predicted Salary : {model.predict(new_employee)[0]:.2f}")
```

**5. Logistic Regression – Student pass/fail**

```python

from sklearn.linear_model import LogisticRegression

# Input Feature (X) - Study hours
X = [[1], [2], [3], [4], [5], [6], [7], [8]]

# Target Value (y) - 0 = Fail, 1 = Pass
y = [0, 0, 0, 0, 1, 1, 1, 1]

# Create a Logistic Regression model
model = LogisticRegression()

# Train the model using study hours and results
model.fit(X, y)

# Predict result for 4 study hours
hours = [[4]]

# [0] accesses the first predicted result
hours_prediction = model.predict(hours)[0]

# Display Pass or Fail based on prediction
print("Result:", "Pass" if hours_prediction == 1 else "Fail")

# [0] accesses the first prediction and [1] gets Pass probability
print(f"Pass Probability : {model.predict_proba(hours)[0][1] * 100 :.2f}%")
```

**6. Logistic Regression – Customer purchase**

```python

# Import NumPy library
import numpy as np

# Import Logistic Regression algorithm
from sklearn.linear_model import LogisticRegression

# Input Features (X) - Age and Annual Income (in thousands)
X = [[20 , 20] , [25 , 25] , [30 , 35] , [35 , 30] , [40 , 60] ,
     [45 , 70] , [50 , 80] , [55 , 90]]

# Target Values (y) - 0 = No Purchase, 1 = Purchase
y = [0 , 0 , 0 , 0 , 1 , 1 , 1 , 1]

# Create a Logistic Regression model
model = LogisticRegression()

# Train the model using input features and target values
model.fit(X, y)

# Predict purchase result for a new customer (Age = 42, Income = 65)
customer = model.predict([[42, 65]])

# Display the predicted result (True = Purchase, False = No Purchase)
print("Will Purchase :", bool(customer[0]))

# Display the probability of purchasing the product
print(f"Purchase Probability : {model.predict_proba([[42, 65]])[0][1] * 100 :.2f}")
```

**7. Multiclass Classification – Iris dataset**


```python

# Import function to load the Iris dataset
from sklearn.datasets import load_iris

# Import function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import Random Forest Classification algorithm
from sklearn.ensemble import RandomForestClassifier

# Import function to calculate model accuracy
from sklearn.metrics import accuracy_score

# Load the Iris dataset
iris = load_iris()

# Split the dataset into 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)

# Create a Random Forest model
model = RandomForestClassifier(random_state=42)

# Train the model using training data
model.fit(X_train, y_train)

# Predict the flower classes using testing data
pred = model.predict(X_test)

# Display the accuracy of the model
print("Accuracy:", accuracy_score(y_test, pred))

# Predict the class of a new flower using its four features
new_flower = [[5.1, 3.5, 1.4, 0.2]]

# Display the predicted flower species
print("Predicted Class:", iris.target_names[model.predict(new_flower)[0]])
```

**9. KNN Classifier using Iris**

```python
# Import function to load the Iris dataset
from sklearn.datasets import load_iris

# Import function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import StandardScaler to scale the feature values
# It's important becoz KNN uses distances to find nearby data points.
from sklearn.preprocessing import StandardScaler

# Import KNN classification algorithm
from sklearn.neighbors import KNeighborsClassifier

# Import make_pipeline to combine scaling and classification
# It combines multiple steps into one pipeine
from sklearn.pipeline import make_pipeline

# Import function to calculate model accuracy
from sklearn.metrics import accuracy_score

# Load the Iris dataset
iris = load_iris()

# Split the dataset into 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, random_state=42, test_size=0.2
)

# Create a pipeline with StandardScaler and KNN classifier
# n_neighbors = 5 means the model considers 5 nearest neighbors
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors = 5))

# Train the model using training data
model.fit(X_train, y_train)

# Predict flower classes using testing data
predict = model.predict(X_test)

# Display the accuracy of the model
print("Accuracy:", accuracy_score(y_test, predict))

# Predict the species of a new flower and display its name
print("Prediction:", iris.target_names[model.predict([[5, 3, 1.5, 0.2]])[0]])
```

**10. KNN – Person category (sample dataset)**

```python
# Import StandardScaler to scale the feature values
from sklearn.preprocessing import StandardScaler

# Import KNN classification algorithm
from sklearn.neighbors import KNeighborsClassifier

# Import make_pipeline to combine scaling and classification
from sklearn.pipeline import make_pipeline

# Define input features: height (cm) and weight (kg)
X = [[150, 45], [155, 50], [160, 55], [170, 70], [175, 75],
     [180, 80]]

# Define categories: 0 and 1
y = [0, 0, 0, 1, 1, 1]

# Create a pipeline with StandardScaler and KNN classifier
# n_neighbors = 3 means the model considers 3 nearest neighbors
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=3))

# Train the model using the input features and categories
model.fit(X, y)

# Define the height and weight of a new person
person = [[172, 72]]

# Predict the category of the new person
predict = model.predict(person)

# Display the predicted category
print("Predicted Category :", predict[0])
```

**13. Bayesian Network – Possibility of rain (simple probability model)**

```python
# Define weather conditions
cloudy = True
high_humidity = True
low_pressure = True

# Set the initial probability of rain to 20%
rain_probability = 0.2

# Increase rain probability if the sky is cloudy
if cloudy:
    rain_probability = rain_probability + 0.2

# Increase rain probability if humidity is high
if high_humidity:
    rain_probability += 0.3

# Increase rain probability if air pressure is low
if low_pressure:
    rain_probability = rain_probability + 0.1

# Make sure the probability does not exceed 1 (100%)
rain_probability = min(rain_probability, 1)

# Display the rain probability in percentage format
print(f"Probability of rain: {rain_probability * 100:.2f}%")

# Predict rain if probability is greater than 50%
print("Prediction :", "Rain" if rain_probability > 0.5 else "No Rain")
```

**14. Bayesian Network – Disease based on symptoms (simple model)**

```python
# Define the symptoms
fever = True
cough = True
fatigue = False

# Set the initial disease probability to 10%
probability = 0.10

# Increase probability if the person has fever
if fever:
    probability = probability + 0.4

# Increase probability if the person has cough
if cough:
    probability = probability + 0.3

# Increase probability if the person has fatigue
if fatigue:
    probability = probability + 0.2

# Make sure probability does not exceed 1 (100%)
probability = min(probability, 1)

# Display the disease probability in percentage
print(f"Probability of having the disease: {probability * 100:.2f}%")

# Predict disease if probability is greater than 50%
print("Prediction:", "Disease" if probability > 0.5 else "No Disease")
```

**15. Decision Tree classifier using Iris**

```python
# Import function to load the Iris dataset
from sklearn.datasets import load_iris

# Import Decision Tree classifier and tree visualization function
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Import Matplotlib for displaying the Decision Tree diagram
import matplotlib.pyplot as plt

# Load the Iris dataset
iris = load_iris()

# Create a Decision Tree model with a maximum depth of 3
model = DecisionTreeClassifier(random_state=42, max_depth=3)

# Train the model using all the Iris data and target values
model.fit(iris.data, iris.target)

# Define the measurements of a new flower
sample = [[5.1, 3.5, 1.4, 0.2]]

# Predict the category of the new flower
predict = model.predict(sample)

# Display the predicted flower species
print("Flower:", iris.target_names[predict[0]])

# Draw the Decision Tree with feature names and flower names
# filled=True adds colors to the tree nodes
plot_tree(model, feature_names=iris.feature_names,
          class_names=iris.target_names, filled=True)

# Display the Decision Tree diagram
plt.show()
```

**16. Decision Tree – Student Pass/Fail**

```python
# Import Decision Tree classification algorithm
from sklearn.tree import DecisionTreeClassifier

# Define input features: study hours and attendance percentage
X = [[1, 50], [2, 60], [3, 65], [4, 70], [5, 80], [6, 90]]

# Define output categories: 0 = Fail, 1 = Pass
y = [0, 0, 0, 1, 1, 1]

# Create a Decision Tree classification model
model = DecisionTreeClassifier(random_state=42)

# Train the model using the input features and output categories
model.fit(X, y)

# Define the study hours and attendance of a new student
student = [[4, 75]]

# Predict the result of the student
# [0] extracts the predicted category from the prediction array
predict = model.predict(student)[0]

# Display the result based on the predicted category
print("Result:", "Pass" if predict == 1 else "Fail")
```

**17. SVM classifier using Iris**

```python
# Import function to load the Iris dataset
from sklearn.datasets import load_iris

# Import StandardScaler to scale the feature values
from sklearn.preprocessing import StandardScaler

# Import function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import Support Vector Machine classification algorithm
from sklearn.svm import SVC

# Import make_pipeline to combine scaling and classification
from sklearn.pipeline import make_pipeline

# Import function to calculate model accuracy
from sklearn.metrics import accuracy_score

# Load the Iris dataset
iris = load_iris()

# Split the dataset into 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, random_state=42, test_size=0.2
)

# Create a pipeline with StandardScaler and SVC
# StandardScaler scales the features, and SVC performs classification
model = make_pipeline(StandardScaler(), SVC())

# Train the model using the training data
model.fit(X_train, y_train)

# Predict flower categories using the testing data
predict = model.predict(X_test)

# Define the measurements of a new flower
sample = [[5.1, 3.5, 1.4, 0.2]]

# Display the accuracy of the model
print("Accuracy:", accuracy_score(y_test, predict))

# Predict the species of the new flower and display its name
print("Flower:", iris.target_names[model.predict(sample)[0]])
```
