**1. Simple Linear Regression – House price based on area**

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Area in square feet and price in lakh

# Input Feature (X) is in 2D format
X = np.array([[500], [700], [900], [1100], [1300]])

# Target Value (y) is in 1D format
y = np.array([5, 7, 9, 11, 13])

# Create a Linear Regression model
model = LinearRegression()

# Train the model using input and target values
model.fit(X, y)

# Area for which we want to predict the price
area = [[1000]]

# [0] accesses the first predicted value from the array
print("Predicted house price:", model.predict(area)[0], "lakh")

# [0] accesses the first slope value (only one input feature)
print("Slope:", model.coef_[0])

# Displays the intercept value (no index needed)
print("Intercept:", model.intercept_)
```

**9. KNN Classifier using Iris**

```python
# Import function to load the Iris dataset
from sklearn.datasets import load_iris

# Import function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import StandardScaler to scale the feature values
# It's important becoz KNN uses distances to find nearby data points.
from sklearn.preprocessing import StandardScaler

# Import KNN classification algorithm
from sklearn.neighbors import KNeighborsClassifier

# Import make_pipeline to combine scaling and classification
# It combines multiple steps into one pipeine
from sklearn.pipeline import make_pipeline

# Import function to calculate model accuracy
from sklearn.metrics import accuracy_score

# Load the Iris dataset
iris = load_iris()

# Split the dataset into 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, random_state=42, test_size=0.2
)

# Create a pipeline with StandardScaler and KNN classifier
# n_neighbors = 5 means the model considers 5 nearest neighbors
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors = 5))

# Train the model using training data
model.fit(X_train, y_train)

# Predict flower classes using testing data
predict = model.predict(X_test)

# Display the accuracy of the model
print("Accuracy:", accuracy_score(y_test, predict))

# Predict the species of a new flower and display its name
print("Prediction:", iris.target_names[model.predict([[5, 3, 1.5, 0.2]])[0]])
```

**13. Bayesian Network – Possibility of rain (simple probability model)**

```python
# Define weather conditions
cloudy = True
high_humidity = True
low_pressure = True

# Set the initial probability of rain to 20%
rain_probability = 0.2

# Increase rain probability if the sky is cloudy
if cloudy:
    rain_probability = rain_probability + 0.2

# Increase rain probability if humidity is high
if high_humidity:
    rain_probability += 0.3

# Increase rain probability if air pressure is low
if low_pressure:
    rain_probability = rain_probability + 0.1

# Make sure the probability does not exceed 1 (100%)
rain_probability = min(rain_probability, 1)

# Display the rain probability in percentage format
print(f"Probability of rain: {rain_probability * 100:.2f}%")

# Predict rain if probability is greater than 50%
print("Prediction :", "Rain" if rain_probability > 0.5 else "No Rain")
```

**14. Bayesian Network – Disease based on symptoms (simple model)**

```python
# Define the symptoms
fever = True
cough = True
fatigue = False

# Set the initial disease probability to 10%
probability = 0.10

# Increase probability if the person has fever
if fever:
    probability = probability + 0.4

# Increase probability if the person has cough
if cough:
    probability = probability + 0.3

# Increase probability if the person has fatigue
if fatigue:
    probability = probability + 0.2

# Make sure probability does not exceed 1 (100%)
probability = min(probability, 1)

# Display the disease probability in percentage
print(f"Probability of having the disease: {probability * 100:.2f}%")

# Predict disease if probability is greater than 50%
print("Prediction:", "Disease" if probability > 0.5 else "No Disease")
```

**19. Sentiment analysis – Movie reviews**

```python
# Import CountVectorizer to convert text into numerical word counts
from sklearn.feature_extraction.text import CountVectorizer

# Import Multinomial Naive Bayes classification algorithm
from sklearn.naive_bayes import MultinomialNB

# Define sample movie reviews
reviews = [
    "This movie is excellent and amazing",
    "I loved this wonderful movie",
    "This movie is boring and bad",
    "I hated this terrible movie"
]

# Define sentiment labels for each review
labels = ["Positive", "Positive", "Negative", "Negative"]

# Create a CountVectorizer object
vectorizer = CountVectorizer()

# Convert text reviews into numerical word-count features
# fit_transform() learns the vocabulary and converts the reviews into numbers
X = vectorizer.fit_transform(reviews)

# Create a Multinomial Naive Bayes model
model = MultinomialNB()

# Train the model using numerical reviews and their sentiment labels
model.fit(X, labels)

# Define a new movie review for sentiment prediction
new_reviews = ["This movies is amazing and wonderful"]

# Convert the new review into numerical features
# transform() uses the vocabulary learned from the training reviews
new_X = vectorizer.transform(new_reviews)

# Predict the sentiment of the new review
# [0] extracts the prediction for the first review
predict = model.predict(new_X)[0]

# Display the predicted sentiment
print("Sentiment:", predict)
```

**20. Sentiment analysis – Customer reviews**

```python
# Import CountVectorizer to convert text into numerical word counts
from sklearn.feature_extraction.text import CountVectorizer

# Import Logistic Regression classification algorithm
from sklearn.linear_model import LogisticRegression

# Define sample product reviews
reviews = [
    "great product and good quality",
    "excellent service",
    "bad product and poor quality",
    "very disappointing"
]

# Define sentiment labels: 1 = Positive, 0 = Negative
labels = [1, 1, 0, 0]

# Create a CountVectorizer object
vectorizer = CountVectorizer()

# Convert text reviews into numerical word-count features
X = vectorizer.fit_transform(reviews)

# Create a Logistic Regression model
model = LogisticRegression()

# Train the model using numerical reviews and sentiment labels
model.fit(X, labels)

# Define a new review for sentiment prediction
new_reviews = ["good quality and excellent product"]

# Convert the new review into numerical features and predict its sentiment
# [0] extracts the prediction for the first review
result = model.predict(vectorizer.transform(new_reviews))[0]

# Display the predicted sentiment based on the result
print("Sentiment:", "Positive" if result else "Negative")
```

**21. Simple Linear Regression – evaluate MAE, MSE, RMSE, R²**

```python
# Import NumPy for creating numerical arrays
import numpy as np

# Import Linear Regression algorithm
from sklearn.linear_model import LinearRegression

# Import evaluation metrics for regression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Define input data (X)
X = np.array([[1], [2], [3], [4], [5], [6]])

# Define actual output values (y)
y = np.array([10, 20, 30, 40, 50, 60])

# Create a Linear Regression model
model = LinearRegression()

# Train the model using input and output data
model.fit(X, y)

# Predict output values using the input data
pred = model.predict(X)

# Calculate and display Mean Absolute Error (MAE)
print("Mean Absolute Error:", mean_absolute_error(y, pred))

# Calculate and display Mean Squared Error (MSE)
print("Mean Squared Error:", mean_squared_error(y, pred))

# Calculate and display Root Mean Squared Error (RMSE)
print("Root Mean Squared Error:", np.sqrt(mean_squared_error(y, pred)))

# Calculate and display R-squared score (R2)
print("R2 Score:", r2_score(y, pred))
```

**23. Multivariate Linear Regression – feature selection and evaluation**

```python
# Import function to load the Diabetes dataset
from sklearn.datasets import load_diabetes

# Import function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import SelectKBest for feature selection and f_regression to score features
from sklearn.feature_selection import SelectKBest, f_regression

# Import evaluation metrics for regression
from sklearn.metrics import r2_score, mean_squared_error

# Import Linear Regression algorithm
from sklearn.linear_model import LinearRegression

# Import NumPy for numerical operations
import numpy as np

# Load the Diabetes dataset
diabetes = load_diabetes()

# Split the dataset into 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    diabetes.data, diabetes.target, random_state=42, test_size=0.2
)

# Create a feature selector to select the 5 best features
selector = SelectKBest(score_func=f_regression, k=5)

# Select the 5 best features from the training data
X_train_sel = selector.fit_transform(X_train, y_train)

# Apply the same feature selection to the testing data
X_test_sel = selector.transform(X_test)

# Create a Linear Regression model
model = LinearRegression()

# Train the model using the selected training features
model.fit(X_train_sel, y_train)

# Predict output values using the selected testing features
predict = model.predict(X_test_sel)

# Display the names of the 5 selected features
# get_support() identifies selected features using True and False
print("Selected Features:", np.array(diabetes.feature_names)[selector.get_support()])

# Calculate and display the R2 score
print(f"R2 Score: {r2_score(y_test, predict):.2f}")

# Calculate and display Root Mean Squared Error (RMSE)
print(f"Root Mean Squared Error: {np.sqrt(mean_squared_error(y_test, predict)):.2f}")
```

**24. House price regression – impact of individual features**
