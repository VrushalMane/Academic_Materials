**1. Simple Linear Regression – House price based on area**

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Area in square feet and price in lakh

# Input Feature (X) is in 2D format
X = np.array([[500], [700], [900], [1100], [1300]])

# Target Value (y) is in 1D format
y = np.array([5, 7, 9, 11, 13])

# Create a Linear Regression model
model = LinearRegression()

# Train the model using input and target values
model.fit(X, y)

# Area for which we want to predict the price
area = [[1000]]

# [0] accesses the first predicted value from the array
print("Predicted house price:", model.predict(area)[0], "lakh")

# [0] accesses the first slope value (only one input feature)
print("Slope:", model.coef_[0])

# Displays the intercept value (no index needed)
print("Intercept:", model.intercept_)
```

**2. Simple Linear Regression – Marks based on study hours**

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Input Feature (X) - Study hours in 2D format
X = np.array([[1], [2], [3], [4], [5]])

# Target Value (y) - Marks in 1D format
y = np.array([20, 35, 50, 65, 80])

# Create a Linear Regression model
model = LinearRegression()

# Train the model using study hours and marks
model.fit(X, y)

# Predict marks for 6 study hours
study_hour = [[6]]

# [0] accesses the first predicted value
print("Predicted marks:", model.predict(study_hour)[0])
```

**3. Multivariate Linear Regression – House price**

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create a dataset using Pandas DataFrame
data = pd.DataFrame({
    "Area": [600, 800, 1000, 1200, 1500],
    "Bedrooms": [1, 2, 2, 3, 3],
    "Age": [10, 8, 5, 4, 2],
    "Price": [30, 42, 55, 68, 85]
})

# Select input features (X)
X = data[["Area", "Bedrooms", "Age"]]

# Select target value (y)
y = data[["Price"]]

# Create a Multiple Linear Regression model
model = LinearRegression()

# Train the model using the dataset
model.fit(X, y)

# Predict price for a new house
new_house = pd.DataFrame({
    "Area": [1200],
    "Bedrooms": [4],
    "Age": [6]
})

# [0] accesses the first predicted value
print("Predicted price:", model.predict(new_house)[0])
```

**4. Multivariate Linear Regression – Employee salary**

```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Create a dataset using Pandas DataFrame
data = pd.DataFrame({
    "Experience": [2, 3, 4, 5, 6],
    "Education": [9, 12, 16, 18, 20],
    "Age": [25, 30, 35, 40, 45],
    "Salary": [50000, 60000, 70000, 80000, 90000]
})

# Select input features (X)
X = data[["Experience", "Education", "Age"]]

# Select target value (y)
y = data["Salary"]

# Create a Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Create data for a new employee
new_employee = pd.DataFrame({
    "Experience": [5],
    "Education": [15],
    "Age": [45]
})

# Predicts Salary for new Employee
print(f"Predicted Salary : {model.predict(new_employee)[0]:.2f}")
```

**5. Logistic Regression – Student pass/fail**

```python

from sklearn.linear_model import LogisticRegression

# Input Feature (X) - Study hours
X = [[1], [2], [3], [4], [5], [6], [7], [8]]

# Target Value (y) - 0 = Fail, 1 = Pass
y = [0, 0, 0, 0, 1, 1, 1, 1]

# Create a Logistic Regression model
model = LogisticRegression()

# Train the model using study hours and results
model.fit(X, y)

# Predict result for 4 study hours
hours = [[4]]

# [0] accesses the first predicted result
hours_prediction = model.predict(hours)[0]

# Display Pass or Fail based on prediction
print("Result:", "Pass" if hours_prediction == 1 else "Fail")

# [0] accesses the first prediction and [1] gets Pass probability
print(f"Pass Probability : {model.predict_proba(hours)[0][1] * 100 :.2f}%")
```

**6. Logistic Regression – Customer purchase**

```python

# Import NumPy library
import numpy as np

# Import Logistic Regression algorithm
from sklearn.linear_model import LogisticRegression

# Input Features (X) - Age and Annual Income (in thousands)
X = [[20 , 20] , [25 , 25] , [30 , 35] , [35 , 30] , [40 , 60] ,
     [45 , 70] , [50 , 80] , [55 , 90]]

# Target Values (y) - 0 = No Purchase, 1 = Purchase
y = [0 , 0 , 0 , 0 , 1 , 1 , 1 , 1]

# Create a Logistic Regression model
model = LogisticRegression()

# Train the model using input features and target values
model.fit(X, y)

# Predict purchase result for a new customer (Age = 42, Income = 65)
customer = model.predict([[42, 65]])

# Display the predicted result (True = Purchase, False = No Purchase)
print("Will Purchase :", bool(customer[0]))

# Display the probability of purchasing the product
print(f"Purchase Probability : {model.predict_proba([[42, 65]])[0][1] * 100 :.2f}")
```

**7. Multiclass classification – Iris dataset**


```python

# Import function to load the Iris dataset
from sklearn.datasets import load_iris

# Import function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import Random Forest Classification algorithm
from sklearn.ensemble import RandomForestClassifier

# Import function to calculate model accuracy
from sklearn.metrics import accuracy_score

# Load the Iris dataset
iris = load_iris()

# Split the dataset into 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)

# Create a Random Forest model
model = RandomForestClassifier(random_state=42)

# Train the model using training data
model.fit(X_train, y_train)

# Predict the flower classes using testing data
pred = model.predict(X_test)

# Display the accuracy of the model
print("Accuracy:", accuracy_score(y_test, pred))

# Predict the class of a new flower using its four features
new_flower = [[5.1, 3.5, 1.4, 0.2]]

# Display the predicted flower species
print("Predicted Class:", iris.target_names[model.predict(new_flower)[0]])
```
